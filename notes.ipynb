## Intro Page:

HHL --> Used to solve matrix operations on a quantum computer

HHL --> QPE --> QPE requires high circuit depth

To get around high circuit depth --> classical-quantum hybrid algorithm is used --> VQE & QAOA


In hybrid problems, a problem is encoded in a **Hermitian matrix**. Here:
An expectation value $<A>$ with respect to an **ansatz state** $\ket{\psi(\theta)}$ is **iteratively optimized** by tuning parameter $\theta$.

**Central idea of hybrid algorithm --> divide into classical part & quantum part.**

### Article's framework --> new hybrid approach --> quantum circuit learning (QCL)

**Goal**: Quantum machine learning with low-depth quantum circuit. Works for both supervised and unsupervised learning.

**How it works**: Feed quantum circuit input data & tune $\theta$ to optimize circuit output. Any quantum circuit driven by QCL can approximate any analytical function given enough qubits.

**Tuning**: Gradient-based parameter optimization is employed.

---

## Algorithm:

**Supervised Learning Stratedgy**: An algorithm with an input set ${x_i}$ and teacher data ${f(x_i)}$ is given. The algorithm learns to output $y_i = y(x_i, \theta)$ close to ${f(x_i)}$ by **tuning $\theta$**. QCL assigns output $y_i$ to a **quantum circuit** (quantum part) and updates parameter $\theta$ to a **classical computer** (classical part). 
**Learning objective**: minimize a cost function --> difference between teacher and output --> changes by tuning $\theta$. Quadratic cost example:

$$ \boxed{\; L = \Sigma_i \; || f(x_i) - y_i || ^ 2 \;} $$

**Unsupervised Learning Stratedgy**: Like supervised learning, but without a teacher or an input set. Instead, only input data is provided, and some cost function not involving the teacher is minimized.

---

### Sumarizing the QCL Algorithm:

#### Step 1:
Encode input data ${x_i}$ as a quantum state (hermitian matrix) $\ket{\psi_{in}(x_i)}$ by applying a unitary input gate $U(x_i)$ to initialized qubits $\ket{0}$.

#### Step 2:
Apply $\theta$-parameterized unitary $U(\theta)$ to input state, and get output state $\ket{\psi_{out}(x_i, \theta)} = U(\theta)\ket{\psi_in(x_i)}$.

#### Step 3:
Measure respective expectation values. Here the subset of Pauli operators $ {B_j} \subset {I, X, Y, Z}^{\otimes N} $ is used. With output function $F$ an output can be defined as $y_i = y(x_i, \theta) \equiv F({<B_j(x_i, \theta)>})$.

#### Step 4:
Minimize cost function $L(f(x_i), y(x_i, \theta))$ of teacher $f(x_i)$ and output $y_i$, by tuning $\theta$ iteratively.

#### Step 5:
Evaluate performance by checking cost function with respect to data set taken independent of training dataset.

---

## Relation to Existing Algorithms:

With a **high-depth, HHL-based** algorithm, quadratic cost can be minimized. This approach is 'matrix inversion' and is similar to **quantum support vector machines**. 

In opposition, **QCL** is **low-depth**, and is applied to regression problems to minimize the cost by **iterative optimization**.

### Quantum Reservoir Computing:
**Quantum reservoir computing** (QRC) is similar, in passing the central optimization scheme to a classical computer. In QRC, the output is defined as $y(x_i) \equiv w \; \cdot <B>$ where $B$ is a set of observables taken from quantum many-body dynamics driven with a fixed Hamiltonian and $w$, the weight vector, is tuned on a classical device to minimize the cost function.

This idea originates from the 'echo-state network approach'. Let QRC be the **quantum** echo-state network, then **QCL**--which tunes the network--is the quantum version of a **basic neural network**.

In QVE & QAOA, the weighted sum of measured expectation values $w_{fixed}\; \cdot <B(\theta)>$ is minimized by tuning $\theta$. **For example**, let $x$ be the input of a problem--calculating the geometry of a molecule/topology of a graph--which is encoded in the weight vector $w_{fixed}$ as $w_{fixed}(x)$. Here is a special case in which unitary $U(x)$ and cost function $L = w_{fixed} \; \cdot <B>$ is not used.

## QCL to Approximate a Function:

Consider a simple case where input data is 1D. For any dimension, the following argument stands. Let $x$ be an input datum, and $\rho_{in}(x) = \ket{\psi_{in}(x)} \bra{\psi_{in}(x)}$ be the corresponding density operator. Then, $\rho_{in}$ can be expanded by a set of Pauli operators ${P_k} = {I, X, Y, Z}^{\otimes N}$ with $a_k(x)$ as coefficients $\rho_{in}(x) = \Sigma_k a_k(x) P_k $. A parameterized unitary transformation $U(\theta)$ acting on $\rho_{in}$ creates the output state. It follows that the output can also be expanded by ${P_k}$ with ${b_k(x, \theta)}$. Now let $u_{ij}(\theta)$ be such that $b_m(x, \theta) = \Sigma_k u_{mk}(\theta) a_k(x)$. Then, $b_m$ is the expectation value of the Pauli ovservable. Thus, the output is a linear combination of input coefficient functions $a_k$ under unitarity constrains imposed on ${u_{ij}}$. 

### The relations below are given to show classical vs hybrid algorithm equivalency:

- VQE/QAOA <--> Simulated Annealing
$$ w_{fixed} \; \cdot <B(\theta)> $$ 
to be minimized via $\theta$.
- QRC <--> Echo-State Network
$$ <B(x)> \rightarrow || f(x) - w \; \cdot <B(x)> || $$ 
to be minimized via $w$.
- QCL <--> Neural Network
$$ <B(x, \theta)> \rightarrow L(f(x), <B(x, \theta)>) $$ 
to be minimized via $\theta$.

When the teacher $f(x)$ is an analytical function, it is shown that QCL can approximate the function by a simple case with an input state generated by single-qubit rotations. The tensor product structure of the quantum system is used to consider a state of $N$ qubits:

$$ \rho_{in}(x) = \dfrac{1}{2^N} \Sigma^N_{i = 1} [I + xX_i + \sqrt{1 - x^2} Z_i] $$

This state can be created for any $x e [-1, 1]$ using single qubit rotations, like 

$$\Pi^N_{i = 1} R^Y_i (sin^{-1}x)$$

where $R^Y_i(\phi)$ is the rotation of the $i$ th qubit around the $y$ axis with angle $\phi$.

A unitary transformation on $\rho_{in}$ generates a respective $N$th-order polynomial as the expetation value of an observable. For example, a term like $x \sqrt{(1 - x^2)}$ can enhance its ability to approximate a function. Note that the highest-order term $x^N$ is hidden in an observable $X^{\otimes N}$, and extracting $x^N$ means transfering the nonlocal observable $X^{\otimes N}$ to a single-qubit observable using entanglement like CNOT. Entangling nonlocal operations can result in non linear outputs.

#### For multi-dimensional inputs:

Given $d$-dimensional data, $x = {x_1, x_2, ..., x_d}, to get higher terms up to the $n_k$th $(k = 1, ..., d)$ for each datum, encode this datum in a $N = \Sigma_k n_k$-qubit quantum state as

$$ \rho_in(x) = \dfrac{1}{2^N} \Sigma^d_{k = 1} (\Sigma^{n_k}_{i = 1} [I + x_kX_i + \sqrt{1 - x^2_k} Z_i]) $$

The unitarity condition $u_{ij}$ eliminates the problem of overfitting, which is critical in performance of the model (in classical ML this is handled with a regularization term in the cost function). 

**For example** in ridge learning, this 'regularization term' is $||w||^2$ in the quadratic cost function

$$ L = \Sigma_i || f(x_i) - w \cdot \phi(x_i) ||^2 + ||w||^2 $$

where $L$ is minimized. In QCL, the weight vector $w$ corresponds to the matrix element $u_{ij}$. Note that the norm of the row vector $||u_i||$ is restricted to unity by unitarity--preventing overfitting.

---

## Quantum Advantage with QCL:

- Nonlinearity of **tensor product** --> approximation of **any analytical function** --> **nonlinear basis functions** are critical for many ML methods

- Precise + complex model = large # of basis functions needed --> **more** basis functions = **more computational cost** of learning --> use **kernel-trick** method

- Kernel-trick method circumbents directly using large numbers of basis functions. Instead of the kernel-trick method, QCL directly uses the exponential # of functions with respect to the # of qubits to model the teacher (**intractable** on classical computers). This intertractableness is a possible **quantum advantage** of QCL (not obvios with VQE & QAOA).

### Potential power of QCL to represent complex functions:

**Potential problem**: Learn the output of QCL--that can use unlimited resources in the learning process--using classical neural networks. Then it has to learn the relations between inputs and outputs of a quantum circuit, which includes universal quantum cellular automata. This kind of problem cannot be solved using a polynomial-sized classical computer, and thus implies QCL has the potential power to represent more complex functions than its classical counterpart.

## Optimization procedure:

### In VQE: 
- **Small** parameter space: gradient-free methods like Nelder-Mead
- **Large** parameter space: gradient-based methods

### In neural networks:
- Backpropagation method (gradient descent) is used in the learning process

To calculate the gradient of an observable's expectation value with respect to circuit parameter $\theta$, let unitary $U(\theta)$ consist of a chain of unitary transformations

$$ \Pi^l_{j = 1} U_j (\theta_j) $$

on a state $\rho_{in}$ and $B$ is the observable to be measured. Let $U_{j:k} = U_j ... U_K$ for simplicity. Then, $<B(\theta)> is given as

$$ <B(\theta)> = Tr(B U_{l:1 \rho_{in} U^{\dag} _{l:1}}) $$

Assume $U_j$ is generated by a Pauli product $P_j$, so

$$ U_j(\theta) = e^{-i \theta_j P_j / 2} $$

Then, the gradient is calculated to be

$$ \dfrac{\delta <b>}{\delta \theta_j} = - \dfrac{i}{2} Tr(BU_{l:j} [P_j, U_{j-1:1 \rho_{in}} U^{\dag}_{j - 1:1}] U^{\dag}_{l:j}) $$

Using the commutator property for an arbitrary operator $rho$, the gradient on the quantum circuit is computed as

$$ [P_j, \rho] = i \biggr[ U_j \biggr( \dfrac{\pi}{2} \biggr) \rho U^{\dag}_j \biggr( \dfrac{\pi}{2} \biggr) - U_j \biggr(- \dfrac{\pi}{2} \biggr) \rho U^{\dag}_j \biggr( - \dfrac{\pi}{2} \biggr) \biggr] $$

The gradient in its final form is then

$$ \dfrac{\delta <B>}{\delta \theta_j} = \dfrac{1}{2} Tr \biggr[ B U_{l:j + 1} U_j \biggr(\dfrac{\pi}{2} \biggr) \rho_j U^{\dag}_{j} \biggr( \dfrac{\pi}{2} \biggr) U^{\dag}_{l:j+1} \biggr] - \dfrac{1}{2} Tr \biggr[ B U_{l:j + 1} U_j \biggr( - \dfrac{\pi}{2} \biggr) \rho_j U^{\dag}_{j} \biggr( - \dfrac{\pi}{2} \biggr) U^{\dag}_{l:j+1} \biggr] $$

Here, $\rho_j = U_{j:1} \rho_{in} U^{\dag}_{j:1}$. With a $\pm \dfrac{\pi}{2}$ rotation generated by $P_j$ and measuring the output expectation values $<B>^{\pm}_j$ the gradient of the observable $<B>$ can be evaluated.

## Numerical Simulations:

Let $N = 6$ and $D = 6$, and $U(\theta^{(i)}_j)$ be an arbirary single qubit rotation gate with decomposition

$$ U(\theta^{(i)}_j) = R\theta^{X}_j(\theta^{(i)}_{j1}) \; R\theta^{Z}_j(\theta^{(i)}_{j1}) \; R\theta^{X}_j(\theta^{(i)}_{j3}) $$

$H$ is the Hamiltonian of a fully connected transverse **Ising model**

$$ H + \Sigma^N_{j = 1} a_j X_j + \Sigma^N_{j = 1} \Sigma^{j - 1}_{k = 1} J_{jk} Z_j Z_k $$

where $a_j$ and $J_{jk}$ are taken randomly from a uniform districution on [-1, 1]. The evolution time $T$ is constant at 10. This hamiltonian generates a highly entangled state, and is ineffective on classical computers.

Initially, $\theta$ is initialized with random numbers uniformly distributed on $[0, 2 \pi]$. Outputs are taken from $Z$ expectation values of the first qubits. 

- Functions $f(x) = x^2, f(x) = e^x, sin(x), |x|$ are fitted to demonstrate nonlinearity. 
- Cost function is a **quadratic loss**
- Number of teacher samples is 100
- Input state $\rho_{in}(x)$ is prepared by appying $U_{in}(x) = \Pi_j R^Z_j(cos^{-1}x^2) R^Y_J(sin^{-1}x)$ to qubits initialized in $\ket{0}$.

The quantum circuit is trained to classify (into classes 0 or 1) based on each input training data point $x_i = (x_{i, 0}, x_{i, 1})$. The teacher $f(x_i)$ for each input $x_i$ is a 2D vector (1, 0) for class 0 and (0, 1) for class 1. The total # of teacher samples is 100 for class 0 + 100 for class 1 = 200 total. This value is taken from the expectation of the first two qubits of the Pauli Z operator, and they are transformed by the softmax function $F$. 

For a $d$-dimensional vector $q$, the softmax function returns $d$-dimensional vector $F(q)$, where its $k$th element is $F_k(q) = e^{qk} / \Sigma_i e^{q_i}$. The output $y_i = (y_{i, 0}, y_{i, 1})$ is defined as $F(<Z_1(x_i, \theta)>, <Z_2(x_i, \theta)>). The cost function is the cross-entropy

$$ L = \Sigma_i \Sigma_{k \mu {0, 1}} (f(x_i))_k log y_{ik} $$

The input state is prepared by applying

$$ U_{in}(x) = \Pi_j R^Z_j(cos^{-1}x^2_{i, j \; mod \; 2}) \; R^Y_j(sin^{-1}x^2_{i, j \; mod \; 2}) $$

to qubits initialized in $\ket{0}$.

#### The quantum vs classical difference: 
The kernel-trick support vector machines **discards direct usage of large numbers of basis functions**, while QCL **utilizes exponentially large numbers of basis functions**, taking advantage of quantum processing advantage. 

Thus, QCL benefits from quantum advantage.

For fitting quantum many-body dynamics, QCL is also used. Here, the dynamics of a 10-spin system under the fully connected transverse Ising Hamiltonian is performed to generate teaching data. Coefficients $a_j$ and $J_{jk}$ are taken from a uniform distribution on [-1, 1]. The dynamics begin from the initialized state $\ket{0}^{\otimes 10}$. For practicality, experimentally obtained dynamics of a quantum system with an unknown hamiltonian can be used as teacher data. 

The learned dynamics are of $Z$ expectation of three spins during $t \mu [T_{transient}, T_{transient} + 8]$, spanning [-1, 1] uniformly by $t = 4(x + 1) + T_{transient}$. 
- The first three qubit $Z$ expectation values are taken as outputs from the quantum circuit.
- Here, quadratic cost function is used
- There are 100 reacher samples
- Multiplication constant $a$ is fixed to unity

- Exponentially large space of a quantum system + nonlinear functions + low-depth = QCL

## Intro Page:

Harware **coherence time limitations** --> mitigate practical implementations of many quantum algorithms 

NISQRC algorithm --> makes **mid-circuit measurements** & **deterministic resets** --> **reduces circuit execution** while maintaining appropriate temporal memory (by Volterra Series analysis)

NISQRC overcomes...
- Limitations of **finite coherence**
- Information scrambling in monitored circuits & sampling noise
- These are problems persisting even on fault-tolerant quantum computers

Channel equalization task --> recovers test signal symbols subject to a distorting channel

Recurrent neural networks --> ML algorithms that handle **temporal/sequential dependencies** --> used in natural language processing (streaming data, edge computing, forecasting, etc.)

**Physical neural networks** (PNNs) uses physical systems to compute a trainable transformation on an input signal. 
- A branch of PNNs --> online data processing --> **physical reservoir computing** --> trainable component = linear projector acting on observable state of the physical system
- Fast optimization via single value decomposition routines --> enables **temporal learning** on hardware
- PNNs for quantum systems offer large potential for scalability, efficient, and faster ML --> **due to evolution taking place in Hilbert space, which scales exponentially with number of nodes**
- Current **QML** on **NISQ** (noisy intermediate-scale quantum hardware) is restricted by **sampling noise** (QSN), **barren plateaus**--regions where optimization is exponentially hard, **long data streams** (fidelity and coherence times of nodes **limit large data size**), and **backaction** (information scrambling & thermalization, making it hard to retrieve data correlated with inputs).

For clasical systems --> **'fading memory property'** is required for physical system to retain temporal memory (doesn't degrade on long data streams) --> restricts design of classical reservoir (and how input data is encoded)

**Volterra series theory** analyzes the memory properties of a classical dynamical system --> doesn't yet exist for quantum systems

### **Goal**: Present a volterra theory for quantum systems
- Accounts for measurement backaction (necesary to endow a quantum system with persistent temporal memory on streaming data)
- Based on thisquantum volterra theory --> propose NISQ Reservoir Computing (NISQRC) algorithm --> leverages mid-circuit measurements to process any signal, unlimited by coherence time of logical qubits
- Allows indefinetly-long input signals --> avoid measurement-induced thermalization at long times under repeated measurements --> intrinsic property of algorithm to survive QSN and barren plateaus
